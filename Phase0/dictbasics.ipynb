```python
student = {"name": "Maya", "grade": "A", "city": "Tokyo"}
```

#### Looping Only Through keys

```python
for k in student:
    print(k)
```

#### Looping Only Through values

```python
for val in student.values():
    print(val)
```

#### Loop through both keys and values at the same time

```python
for k,v in student.items():
    print(k,v)

for ky, vl in enumerate(student, start=1):
    print(ky, vl)
```

```python
prices = {"apple": 1.5, "banana": 0.75, "mango": 3.0}

for fruit, price in prices.items():
    print(fruit, price * 2)
```

```python
students = {
    "s001": {"name": "Maya", "grade": "A", "age": 22},
    "s002": {"name": "Alex", "grade": "B", "age": 24},
    "s003": {"name": "Ravi", "grade": "A", "age": 21}
}

for sid, info in students.items():
    if info["grade"] == "A":
        print(info["name"])
```
