#### Softmax Scaling

```python
import numpy as np

def sMax(x):
    e = np.exp(x)
    return e / e.sum()

# Small Values
sml = np.array([1.0, 2.0, 3.0])
print(sMax(sml))

lrg = np.array([10.0, 20.0, 30.0])
print(sMax(lrg))
```

#### L1 Norm - Manhattan Distance

```python
v = np.array([3,-4])
l1_nrm = np.sum(np.abs(v))
print(l1_nrm)
```

#### Norm 2 — The L2 Norm (Euclidean Distance)


```python
v = np.array([3, -4])
l2_norm = np.linalg.norm(v)
print(l2_norm)  # 5.0
```

#### Norm 3 — The L∞ Norm (Maximum Norm)

```python
v = np.array([3, -4])
linf_norm = np.max(np.abs(v))
print(linf_norm)  # 4
```

```python
import numpy as np

v = np.array([3, -4])

l1 = np.sum(np.abs(v))           # 7
l2 = np.linalg.norm(v)           # 5
linf = np.max(np.abs(v))         # 4

print(f"L1 norm: {l1}")
print(f"L2 norm: {l2}")
print(f"L∞ norm: {linf}")
```

```python
import numpy as np

data = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2.0, 1.6],
    [1.0, 1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

ctr = data - data.mean(axis=0)
covMat = np.cov(ctr.T)
print(covMat)
eigVal, eigvec = np.linalg.eig(covMat)
print("\nEigenvalues:", eigVal)
print("Eigenvectors:\n", eigvec)

ttl = eigVal.sum()
frac = eigVal.max() / ttl
print(f"\nTop direction captures {frac*100:.1f}% of the variation")
```

```python
odr = np.argsort(eigVal)[::-1]
print(odr)
eigVal = eigVal[odr]
eigvec = eigvec[:, odr]
print(eigvec.shape)
print("\nFirst principal component (direction of most spread):")
print(eigvec[:, 0])
print("Its eigenvalue (how much spread along it):", eigVal[0])
```

```python
np.random.seed(42)  # Simulate=ing the dataset that is naturally elongated in 1 dierction
nPts = 200

t = np.random.randn(nPts) * 3   # Spread along the Main Direction
ns = np.random.randn(nPts) * 0.5    # Spread small ppl to it

x = t + ns
y = t - ns

dta = np.column_stack([x,y])

# 1. Center the Data
ctr = dta - dta.mean(axis=0)

# 2. Covariance Matrix
covMat = np.cov(ctr.T)
print("Covaraince Matrix: \n", covMat)

# 3. EigenValues and EigenVectors

eigvl, eigvc = np.linalg.eig(covMat)
odr = np.argsort(eigvl)[::-1]
eigvl = eigvl[odr]
eigvc = eigvc[:, odr]
print("\nEigenvalues (sorted):", eigvl)
print("Principal directions:\n", eigvc)
print("\n")


# 4. How Much variation does each direction Capture??
toVar = eigvl.sum()
for i, val in enumerate(eigvl):
    print(f"Component {i+1} captures {val/toVar*100:.1f}% of Variation")
print('\n')

# 5. Cpature only the First Principal Component
fpc = eigvc[:, 0]
cmp = ctr @ fpc

print(f"Original Data (first 10 points) :\n", dta[:10])
print(f"Compressed to 1 Dimension (first 10 points):\n", cmp)
```
