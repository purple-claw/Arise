#### Softmax Scaling

```python
import numpy as np

def sMax(x):
    e = np.exp(x)
    return e / e.sum()

# Small Values
sml = np.array([1.0, 2.0, 3.0])
print(sMax(sml))

lrg = np.array([10.0, 20.0, 30.0])
print(sMax(lrg))
```

#### L1 Norm - Manhattan Distance

```python
v = np.array([3,-4])
l1_nrm = np.sum(np.abs(v))
print(l1_nrm)
```

#### Norm 2 — The L2 Norm (Euclidean Distance)


```python
v = np.array([3, -4])
l2_norm = np.linalg.norm(v)
print(l2_norm)  # 5.0
```

#### Norm 3 — The L∞ Norm (Maximum Norm)

```python
v = np.array([3, -4])
linf_norm = np.max(np.abs(v))
print(linf_norm)  # 4
```
