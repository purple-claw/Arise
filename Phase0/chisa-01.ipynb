### `dp(x, y)` — Manual dot product

- **Purpose:** Compute the dot product of two equal-length numeric vectors without using NumPy's `dot` function.
- **Signature:** `dp(x, y)` where `x` and `y` are array-like objects of the same shape.
- **Returns:** A single numeric value equal to the sum of element-wise products (i.e., Σ x_i * y_i).
- **Notes:** This implementation uses element-wise multiplication and `sum()`; it demonstrates the dot product calculation used in vector math and linear algebra examples.

```python
m = np.array([3,4,0])
n = np.array([-4,3,0])

def dp(x,y): # Manual Dot Product
    return (x*y).sum()

dop = dp(m,n)
pod = np.dot(m,n)

mNorm = np.linalg.norm(m)
nNorm = np.linalg.norm(n)

cos = dp(m,n) // ((mNorm)*(nNorm))
ang = np.arccos(cos)

print(f"Cos_Theta : {cos}")
print(f"Angle : {ang}")
print(f"Vector M Norm : {mNorm}, Vector N Norm : {nNorm}")
print(dop == pod)
```

### `softMax(z)` — Numerically stable softmax

- **Purpose:** Convert a vector of raw scores (`z`) into a probability distribution that sums to 1.
- **Signature:** `softMax(z)` where `z` is a 1-D NumPy array of logits or scores.
- **Implementation details:** The function subtracts `np.max(z)` for numerical stability before exponentiating, preventing overflow for large values. It returns `exp(z - max) / sum(exp(z - max))`.
- **Attention scaling:** In the example the scores are optionally scaled by `sqrt(d_k)` (i.e., `scores / np.sqrt(d_k)`), which is commonly used in scaled dot-product attention to keep gradients stable when `d_k` is large.
- **Returns:** A 1-D NumPy array of probabilities corresponding to each input score.

```python
import numpy as np

def softMax(z):
    z_stab = z - np.max(z)
    ezp_z = np.exp(z_stab)
    return ezp_z / ezp_z.sum()

d_k = 64
np.random.seed(0)
q = np.random.randn(d_k)
k = np.random.randn(5, d_k)
scores = k @ q
print("Probs without scaling:", softMax(scores))
print("Probs with scaling:", softMax(scores / np.sqrt(d_k)))
```
