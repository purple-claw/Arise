```python
import numpy as np
import math

purple = np.array([128, 0, 128])
purple = purple * 2
sm = 0
for i in purple:
    sm += (i * i)
sm = math.sqrt(sm)
print(sm)
print(np.linalg.norm(purple))
```

```python
import numpy as np

q = np.array([1, 2])
k1 = np.array([1, 2])
k2 = np.array([-1, -2])
k3 = np.array([2, -1])

print(np.dot(q, k1))   # 5
print(np.dot(q, k2))   # -5
print(np.dot(q, k3))   # 0
```

```python
m = np.array([3,4,0])
n = np.array([-4,3,0])

def dp(x,y): # Manual Dot Product
    return (x*y).sum()

dop = dp(m,n)
pod = np.dot(m,n)

mNorm = np.linalg.norm(m)
nNorm = np.linalg.norm(n)

cos = dp(m,n) // ((mNorm)*(nNorm))
ang = np.arccos(cos)

print(f"Cos_Theta : {cos}")
print(f"Angle : {ang}")
print(f"Vector M Norm : {mNorm}, Vector N Norm : {nNorm}")
print(dop == pod)
```
