#### Exercise 1

Create a 1D array of 20 evenly-spaced values between 0 and 100 using np.linspace, then reshape it into a 4x5 matrix.
Using boolean indexing, extract all values greater than 60 from that matrix.
Replace all values less than 30 with 0, directly in the matrix in place.
Using fancy indexing, extract rows 0, 2, and 3.
Transpose the matrix and print its new shape.

```python
import numpy as np

nms = np.linspace(0,100,20)
print(nms)

rNms = nms.reshape(4,5)
print(rNms)
print(rNms.shape)

msk = (rNms > 60)
msk60 = rNms[msk]
print(msk60)

msk = (rNms <= 30)
rNms[msk] = 0
print(rNms)
print(rNms[[0,2,3]])

trNms = rNms.T
print(trNms)
print(trNms.shape)
```
