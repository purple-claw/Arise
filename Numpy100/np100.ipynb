```python
import numpy as np

def sMax(x):
    e = np.exp(x)
    return e / e.sum()

# Small Values
sml = np.array([1.0, 2.0, 3.0])
print(sMax(sml))

lrg = np.array([10.0, 20.0, 30.0])
print(sMax(lrg))
```
